```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

def kmeans_text_clustering(docs, k=2):
    vectorizer = TfidfVectorizer()
    X = vectorizer.fit_transform(docs)

    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)

    clusters = model.labels_
    for i, doc in enumerate(docs):
        print(f"Doc {i} -> Cluster {clusters[i]}: {doc}")

# Example
docs = [
    "machine learning and deep learning are popular",
    "pizza recipes and cooking tips",
    "supervised and unsupervised algorithms",
    "travel destinations and guides",
    "neural networks and AI models"
]
kmeans_text_clustering(docs, k=2)
```
