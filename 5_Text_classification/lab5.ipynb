```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# -------------------------
# 1) Toy dataset
# -------------------------
docs = [
    "The team won the championship match",  # Sports
    "The player scored a goal in the game",  # Sports
    "Government passed a new law today",  # Politics
    "The election results were announced",  # Politics
    "New smartphone model released",  # Tech
    "Advances in artificial intelligence",  # Tech
]
labels = ["Sports", "Sports", "Politics", "Politics", "Tech", "Tech"]
```

```python
# -------------------------
# 2) TF-IDF representation
# -------------------------
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)
y = np.array(labels)
```

```python
# -------------------------
# 3) Models: Train once
# -------------------------
nb = MultinomialNB().fit(X, y)
dt = DecisionTreeClassifier(random_state=0).fit(X, y)
knn = KNeighborsClassifier(n_neighbors=3).fit(X, y)
```

```python
# -------------------------
# 4) Lab Demo Function
# -------------------------
def categorize_text(sample_text):
    q_vec = vectorizer.transform([sample_text])

    print("Sample Text:", sample_text)
    print("\n--- Model Predictions ---")
    print("Naive Bayes:", nb.predict(q_vec)[0])
    print("Decision Tree:", dt.predict(q_vec)[0])
    print("KNN:", knn.predict(q_vec)[0])
```

```python
# -------------------------
# 6) Try it out
# -------------------------
sample_text = "The government introduced new reforms"
categorize_text(sample_text)
```
