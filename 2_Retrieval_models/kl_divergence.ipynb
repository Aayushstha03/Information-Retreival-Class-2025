```python
import math

# Example documents and query
documents = {
    "doc1": "information retrieval is essential",
    "doc2": "broad research is ongoing",
    "doc3": "topic modeling relates to information",
}
query = "information retrieval modeling"

def preprocess(text):
    return text.lower().split()

# Build vocabulary
vocab = set()
for text in documents.values():
    vocab.update(preprocess(text))
vocab.update(preprocess(query))
vocab = sorted(list(vocab))

# Probability distribution for a text
def prob_dist(text, vocab):
    tokens = preprocess(text)
    total = len(tokens)
    # Laplace smoothing to avoid zero probabilities
    return {term: (tokens.count(term) + 1) / (total + len(vocab)) for term in vocab}

# KL divergence
def kl_divergence(P, Q):
    div = 0.0
    for term in P:
        if P[term] > 0:
            div += P[term] * math.log(P[term] / Q[term])
    return div

# Compute query and document distributions
P_query = prob_dist(query, vocab)
scores = {}
for name, text in documents.items():
    P_doc = prob_dist(text, vocab)
    scores[name] = kl_divergence(P_query, P_doc)

# Rank documents (lower KL = more similar)
ranked = sorted(scores.items(), key=lambda x: x[1])
print("\nKL Divergence Ranking (lower is better):")
for name, score in ranked:
    print(f"{name}: {score:.4f}")
```
